```python
# def = define | function_name | (parameters) | colon
def greet(name):
  message = f"Hello,{name}!Welcome"
  return message
# call the function
result = greet("Amina")
print(result)
print(greet("Kamau"))
```

```python
# Function with multiple parameters
def calculate_grade(score, max_score = 100):
  percentage = (score/max_score)*100
  if percentage >= 90:
    grade = "A"
  elif percentage >= 80:
    grade = "B"
  elif percentage >= 70:
    grade = "C"
  else:
    grade = "F"
  return grade

print(calculate_grade(85))
print(calculate_grade(45,50))
```

```python
# Returning multiple values
def analyse_scores(scores):
  average = sum(scores)/len(scores)
  highest = max(scores)
  lowest = min(scores)
  return average, highest, lowest

# Unpack the results
avg, high, low = analyse_scores([88, 72, 95, 63, 51])
print(f"Average:{avg:.1f}, Highest:{high}, Lowest:{low}")
```

```python
# Lambda functions
# Regular function
def double(x):
  return x*2
print(double(5))
# Equivalent lambda
double = lambda x: x*2
print(double(5))
# Common when sorting with a custom key
students = [{"Name:":"Bob","score:":72},{"Name:":"Alice","score:":81},{"Name:":"Roy","score:":90}]
sorted_students = sorted(students, key = lambda student: student["score:"],reverse=True)
print(sorted_students)
```

```python
# Scope - where variables live
total = 0
def add_to_total(amount):
  global total
  total += amount
add_to_total(50)
add_to_total(30)
print(total)


```
